## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: Which pages should a content editor review first for refresh, given limited review capacity?

Who acts, and how: A content editor or SEO strategist reviews the top N pages from my ranked output and decides whether to refresh, expand, protect, prune, or monitor each one.

Cost of a wrong call: A false positive wastes editor time reviewing a page that wasn't actually declining. A false negative means a real decline goes unnoticed while the page keeps losing traffic. Since review time is limited, precision at the top of the list  matters more than generic accuracy.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
while not os.path.exists("data/raw/content_refresh_anonymized.csv") and os.getcwd() != "/":
    os.chdir("..")
print("Working dir:", os.getcwd())

import pandas as pd
import json

res = json.load(open("outputs/model_results.json"))
print("Baseline Precision@50:", res["baseline"]["baseline_precision_at_50"])
print("Random forest Precision@50:", res["models"]["random_forest"]["precision_at_50"])

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_share = df["trend_direction"].str.lower().eq("down").mean()
print(f"\nShare of pages currently declining: {declining_share:.1%}")

has_position = df[df["avg_position"] > 0]
print(f"Pages with valid position data: {len(has_position):,} of {len(df):,}")

stale_declining_visible = (
    (df["days_since_last_update"] >= 180) &
    (df["trend_direction"].str.lower() == "down") &
    (df["impressions_90d"] >= 500)
).mean()
print(f"Share matching 'stale + declining + visible' pattern: {stale_declining_visible:.1%}")
```
